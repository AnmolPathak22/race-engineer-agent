```python
import fastf1
import matplotlib.pyplot as plt

fastf1.Cache.enable_cache("../data/cache")

session = fastf1.get_session(2024, "Monza", "Q")
session.load()

lap = session.laps.pick_drivers("VER").pick_fastest()
car = lap.get_car_data().add_distance()

fig, ax = plt.subplots(3, 1, figsize=(12, 8), sharex=True)
ax[0].plot(car["Distance"], car["Speed"]);    ax[0].set_ylabel("Speed")
ax[1].plot(car["Distance"], car["Throttle"]); ax[1].set_ylabel("Throttle")
ax[2].plot(car["Distance"], car["Brake"]);    ax[2].set_ylabel("Brake")
ax[2].set_xlabel("Distance (m)")
plt.show()
```
